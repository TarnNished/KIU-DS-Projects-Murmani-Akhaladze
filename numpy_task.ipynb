```python
import numpy as np
np.random.seed(42)
```

```python
print("1. Creating Temperature Data Array")
print("-" * 70)

daily_temps = np.random.uniform(-10, 40, (365, 5))

print(f"Array Shape: {daily_temps.shape}")
print(f"Number of Dimensions: {daily_temps.ndim}")
print(f"Data Type: {daily_temps.dtype}")
print(f"Total Elements: {daily_temps.size}")
print(f"\nSample data (first 3 days for all cities):")
print(daily_temps[:3, :])
```

```python
print("\n2. Creating Sales Matrix")
print("-" * 70)

monthly_sales = np.random.randint(1000, 5001, (12, 4))
print(f"Sales Matrix Shape: {monthly_sales.shape}")
print(f"Sales Data:\n{monthly_sales}")
```

```python
print("\n3. Special Arrays")
print("-" * 70)

identity_5x5 = np.eye(5)
print("Identity Matrix (5x5):")
print(identity_5x5)

linear_space = np.linspace(0, 100, 50)
print(f"\nLinear Space Array (50 values from 0 to 100):")
print(f"First 5 values: {linear_space[:5]}")
print(f"Last 5 values: {linear_space[-5:]}")
```

```python
print("\n1. Basic Slicing")
print("-" * 70)

january_temps = daily_temps[:31, :]
print(f"January data shape: {january_temps.shape}")
print(f"January average per city: {january_temps.mean(axis=0)}")
```

```python
summer_temps = daily_temps[151:243, :]
print(f"\nSummer months data shape: {summer_temps.shape}")
print(f"Summer average per city: {summer_temps.mean(axis=0)}")
```

```python
weekend_temps = daily_temps[4::7, :]
print(f"\nWeekend data shape: {weekend_temps.shape}")
print(f"Weekend average per city: {weekend_temps.mean(axis=0)}")
```

```python
print("\n2. Boolean Indexing")
print("-" * 70)

hot_days_mask = np.any(daily_temps > 35, axis=1)
hot_days_count = np.sum(hot_days_mask)
print(f"Days with temperature > 35°C in any city: {hot_days_count}")
print(f"Day indices (first 10): {np.where(hot_days_mask)[0][:10]}")
```

```python
freezing_days_per_city = np.sum(daily_temps < 0, axis=0)
print(f"\nFreezing days (< 0°C) per city:")
for idx, count in enumerate(freezing_days_per_city):
    print(f"  City {idx+1}: {count} days")
```

```python
comfortable_mask = (daily_temps >= 15) & (daily_temps <= 25)
comfortable_days_per_city = np.sum(comfortable_mask, axis=0)
print(f"\nComfortable days (15-25°C) per city:")
for idx, count in enumerate(comfortable_days_per_city):
    print(f"  City {idx+1}: {count} days")
```

```python
temps_below_minus5 = np.sum(daily_temps < -5)
daily_temps[daily_temps < -5] = -5
print(f"\nData cleaning: Replaced {temps_below_minus5} values below -5°C with -5°C")
```

```python
print("\n3. Fancy Indexing")
print("-" * 70)

specific_days = [0, 100, 200, 300, 364]
specific_days_temps = daily_temps[specific_days, :]
print(f"Temperatures on specific days {specific_days}:")
print(specific_days_temps)
```

```python
q1 = daily_temps[0:91, :].mean(axis=0)
q2 = daily_temps[91:182, :].mean(axis=0)
q3 = daily_temps[182:273, :].mean(axis=0)
q4 = daily_temps[273:365, :].mean(axis=0)

quarterly_avg = np.array([q1, q2, q3, q4])
print(f"\nQuarterly averages (4 quarters × 5 cities):")
print(quarterly_avg)
print(f"\nQuarter 1 avg: {q1}")
print(f"Quarter 2 avg: {q2}")
print(f"Quarter 3 avg: {q3}")
print(f"Quarter 4 avg: {q4}")
```

```python
annual_avg = daily_temps.mean(axis=0)
sorted_indices = np.argsort(annual_avg)[::-1]
temps_sorted = daily_temps[:, sorted_indices]

print(f"\nCities rearranged by annual average (descending):")
print(f"Original annual averages: {annual_avg}")
print(f"Sorted indices: {sorted_indices}")
print(f"Sorted annual averages: {annual_avg[sorted_indices]}")
```

```python
print("\n1. Temperature Analysis")
print("-" * 70)

city_means = daily_temps.mean(axis=0)
city_medians = np.median(daily_temps, axis=0)
city_stds = daily_temps.std(axis=0)

print("Statistics for each city:")
for i in range(5):
    print(f"\nCity {i+1}:")
    print(f"  Mean: {city_means[i]:.2f}°C")
    print(f"  Median: {city_medians[i]:.2f}°C")
    print(f"  Std Dev: {city_stds[i]:.2f}°C")
```

```python
hottest_idx = np.unravel_index(daily_temps.argmax(), daily_temps.shape)
hottest_temp = daily_temps[hottest_idx]
print(f"\nHottest day of the year:")
print(f"  Day {hottest_idx[0]+1}, City {hottest_idx[1]+1}: {hottest_temp:.2f}°C")

coldest_idx = np.unravel_index(daily_temps.argmin(), daily_temps.shape)
coldest_temp = daily_temps[coldest_idx]
print(f"\nColdest day of the year:")
print(f"  Day {coldest_idx[0]+1}, City {coldest_idx[1]+1}: {coldest_temp:.2f}°C")
```

```python
temp_ranges = daily_temps.max(axis=0) - daily_temps.min(axis=0)
print(f"\nTemperature range (max - min) for each city:")
for i, temp_range in enumerate(temp_ranges):
    print(f"  City {i+1}: {temp_range:.2f}°C")
```

```python
city_correlation = np.corrcoef(daily_temps.T)
print(f"\nCorrelation matrix between cities:")
print(city_correlation)
print("\nNote: Values close to 1 indicate strong positive correlation")
```

```python
print("\n2. Sales Analysis")
print("-" * 70)

total_per_category = monthly_sales.sum(axis=0)
print("Total sales per product category:")
for i, total in enumerate(total_per_category):
    print(f"  Category {i+1}: ${total:,}")
```

```python
avg_per_category = monthly_sales.mean(axis=0)
print("\nAverage monthly sales per category:")
for i, avg in enumerate(avg_per_category):
    print(f"  Category {i+1}: ${avg:.2f}")
```

```python
monthly_totals = monthly_sales.sum(axis=1)
best_month = monthly_totals.argmax()
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
print(f"\nBest performing month overall:")
print(f"  {months[best_month]} with ${monthly_totals[best_month]:,}")
```

```python
best_category = total_per_category.argmax()
print(f"\nBest performing category overall:")
print(f"  Category {best_category+1} with ${total_per_category[best_category]:,}")
```

```python
print("\n3. Advanced Computations")
print("-" * 70)

window = 7
moving_avg = np.zeros((daily_temps.shape[0] - window + 1, daily_temps.shape[1]))

for city in range(daily_temps.shape[1]):
    moving_avg[:, city] = np.convolve(daily_temps[:, city],
                                       np.ones(window)/window,
                                       mode='valid')

print(f"7-day moving average shape: {moving_avg.shape}")
print(f"Sample moving average (first 5 days, City 1):")
print(moving_avg[:5, 0])
```

```python
z_scores = np.zeros_like(daily_temps)

for city in range(daily_temps.shape[1]):
    mean = daily_temps[:, city].mean()
    std = daily_temps[:, city].std()
    z_scores[:, city] = (daily_temps[:, city] - mean) / std

print(f"\nZ-scores computed for all cities")
print(f"Z-score mean (should be ~0): {z_scores.mean(axis=0)}")
print(f"Z-score std (should be ~1): {z_scores.std(axis=0)}")
```

```python
print("\nPercentiles (25th, 50th, 75th) for each city:")
for city in range(daily_temps.shape[1]):
    p25 = np.percentile(daily_temps[:, city], 25)
    p50 = np.percentile(daily_temps[:, city], 50)
    p75 = np.percentile(daily_temps[:, city], 75)
    iqr = p75 - p25

    print(f"\nCity {city+1}:")
    print(f"  25th percentile: {p25:.2f}°C")
    print(f"  50th percentile (median): {p50:.2f}°C")
    print(f"  75th percentile: {p75:.2f}°C")
    print(f"  IQR (75th - 25th): {iqr:.2f}°C")
```
